## Import libraries

````python
import os

import pandas as pd
import matplotlib.pyplot as plt
import geopandas as gpd
````

## Import countries shapefile

````python
fp = os.path.join('data', 'ne_50m_admin_0_countries', 'ne_50m_admin_0_countries.shp')
countries = gpd.read_file(fp)

# Simplify column names
countries.columns = countries.columns.str.lower()

# Select columns for analysis
countries = countries[['admin', 'type', 'geometry']]

countries.head(3)
````

## Arctic communities by country

We want to calculate the number of Arctic communities by country

````python
# Create data frame with number of communities per country
n_comms = communities.groupby('country').size().reset_index(name = 'n_communities')
````

Goal: merge the `n_comms` and the `countries` data frames:

To merge two data frames they need to have at least one column in common.

Do they have a column in common?

````python
set(countries.columns).intersection(n_comms.columns)
````

````python
common_columns = set(countries.columns).intersection(n_comms.columns)

if len(common_columns) != 0:
    print(f"Common columns: {common_columns}")
else:
    print("No common columns")
````

Check-in:
Make the previous code into a function called `check_common_columnns`, that takes in two data frames as arguments.

````python
def check_common_columns(df1, df2):
    """ 
    Check if there are common columns between two data frames.

    Args:
        df1 (pd.DataFrame): data frame

    
    """
    common_columns = set(df1.columns).intersection(df2.columns)

    if len(common_columns) != 0:
        print(f"Common columns: {common_columns}")
    else:
        print("No common columns")

    return

check_common_columns(countries, n_comms)
````

## Merge data frames

We can use the `country_names` data frame to add the country names into the `n_comms`.

To merge data frames we can use `pandas.merge()` function

```python
output_df = pd.merge(left_df,
                     right_df,
                     how = type_of_join,
                     on = column_to_join)

```

where

- `output_df` = df resulting from merge
- `left_df` = df we have "on the left side"
- `right_df` = df "on the right side"
- `how` = type of join
- `on` = column to join on, must be present on both data frames

````python
check_common_columns(n_comms, country_names)
````

````python
n_comms = pd.merge(n_comms,
                   country_names,
                   how = 'left',
                   on='country')
n_comms
````

````python
arctic_countries = pd.merge(countries,
                            n_comms,
                            how = 'inner',
                            on = 'admin')
arctic_countries
````

````python
# Merge data frames
arctic_countries = countries.merge(n_comms, how='inner', on='admin')
arctic_countries
````

Note: when merging with a geoDataFrame, the geoDataFrame needs to be the left data frame to keep the geometry.

## Review results

````python
print(f"Countries in n_comms: {len(n_comms)}")
print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

````python
# Countries in n_comms without a match in countries
n_comms[~n_comms['admin'].isin(countries['admin'])]
````

````python
# Looking at values of admin column and checking if any of those values contain 'Aland'
countries[countries['admin'].str.contains('Aland')]
````

````python
# Match the name used in the countries data
n_comms['admin'] = n_comms['admin'].replace('Aland Islands', 'Aland')

# Merge again
arctic_countries = countries.merge(n_comms, how='inner', on='admin')

print(f"Countries in arctic_countries: {len(arctic_countries)}")
````

# Chloropleth map

````python
arctic_countries.plot(column='n_communities', # indicates which column to color by
                      legend=True)
````

## Reprojecting

````python
# Reproject to CRS optimized for Arctic region
arctic_countries = arctic_countries.to_crs('epsg:3413')
````

````python
arctic_countries.crs
````

````python
fig, ax = plt.subplots(figsize=(8,6))
ax.axis('off')

arctic_countries.plot(ax=ax,
                      column='n_communities', # indicates which column to color by
                      legend=True,
                      edgecolor="0.6",
                      linewidth=.7,
                      cmap='PuBuGn',
                      legend_kwds={
                          "label": "Number of Communities",
                          "orientation": "horizontal"
                      })

ax.set_title('Communities at or above 55N, 2022\n(at least 10,000 people, 500 in Alaska)')

plt.show()
````
